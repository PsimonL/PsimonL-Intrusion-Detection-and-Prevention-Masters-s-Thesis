```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
morning_attacks = pd.read_csv('Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv')
ma = morning_attacks.copy()

afternoon_attacks = pd.read_csv('Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv')
aa = afternoon_attacks.copy()

print("Wymiary morning_attacks:", ma.shape)
print("Wymiary afternoon_attacks:", aa.shape)

print(ma.columns.tolist())
print(aa.columns.tolist())
```

```python
ma_columns = set(ma.columns)
aa_columns = set(aa.columns)

if ma_columns == aa_columns:
    print("Nazwy kolumn są identyczne w obu DataFrame.\n")
else:
    print("Nazwy kolumn różnią się między DataFrame.")
    print("Kolumny tylko w morning_attacks:", ma_columns - aa_columns)
    print("Kolumny tylko w afternoon_attacks:", aa_columns - ma_columns)

ma.columns = ma.columns.str.strip()
aa.columns = aa.columns.str.strip()

print("Unikalne wartości w kolumnie 'Label' (morning_attacks):")
print(ma['Label'].unique())

print("Unikalne wartości w kolumnie 'Label' (afternoon_attacks):")
print(aa['Label'].unique())
```

```python
all_attacks = pd.concat([ma, aa], ignore_index=True)

all_attacks.to_csv('Thursday.csv', index=False)

data = pd.read_csv('Thursday.csv')
df = data.copy()


# df = pd.read_csv('Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv')
```

```python
label_encoder = LabelEncoder() 
flag = "2D"
if flag ==  "MD":
    label_encoder = LabelEncoder()
    df['Label'] = label_encoder.fit_transform(df['Label'])
    print("Etykiety odpowiadające wartościom liczbowym:")
    for i in range(len(label_encoder.classes_)):
        print(f"{i}: {label_encoder.classes_[i]}")
elif flag == "2D":
    df['Label'] = df['Label'].map(lambda x: 0 if x == "BENIGN" else 1)


print("Unikalne wartości w kolumnie 'Label':")
print(df['Label'].unique())
print(df['Label'].value_counts()) 
```

```python
rows_before = df.shape[0]

df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)

rows_after = df.shape[0]

rows_dropped = rows_before - rows_after
print(f"Liczba usuniętych wierszy: {rows_dropped}")
```
